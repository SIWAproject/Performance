```python
def calculate_columns(df):
    # Definir las columnas necesarias para cada cálculo con los nombres definitivos
    bw_columns = [('BW7', 'GBW_7', 'Fnl_Birds_Day7'),
                  ('BW14', 'GBW_14', 'Fnl_Birds_Day14'),
                  ('BW21', 'GBW_21', 'Fnl_Birds_Day21'),
                  ('BW28', 'GBW_28', 'Fnl_Birds_Day28'),
                  ('BW35', 'GBW_35', 'Fnl_Birds_Day35'),
                  ('BW42', 'GBW_42', 'Fnl_Birds_Day42')]
    
    fcr_columns = [('FCR0_7', 'GFI_0_7d', 'GBW_7', 'GBW_initial', 'Fnl_Birds_Day7'),
                   ('FCR8_14', 'GFI_8_14d', 'GBW_14', 'GBW_initial', 'Fnl_Birds_Day14'),
                   ('FCR15_21', 'GFI_15_21d', 'GBW_21', 'GBW_14', 'Fnl_Birds_Day21'),
                   ('FCR22_28', 'GFI_22_28d', 'GBW_28', 'GBW_21', 'Fnl_Birds_Day28'),
                   ('FCR29_35', 'GFI_29_35d', 'GBW_35', 'GBW_28', 'Fnl_Birds_Day35'),
                   ('FCR36_42', 'GFI_36_42d', 'GBW_42', 'GBW_35', 'Fnl_Birds_Day42')]
    
    cfcr_columns = [('cFCR0_7', 'GFI_0_7d', 'GBW_7', 'GBW_initial', 'Mort_Wt_D7', 'Cull_Wt_D7', 'Fnl_Birds_Day7'),
                    ('cFCR8_14', 'GFI_8_14d', 'GBW_14', 'GBW_7', 'Mort_Wt_D14', 'Cull_Wt_D14', 'Fnl_Birds_Day14'),
                    ('cFCR15_21', 'GFI_15_21d', 'GBW_21', 'GBW_14', 'Mort_Wt_D21', 'Cull_Wt_D21', 'Fnl_Birds_Day21'),
                    ('cFCR22_28', 'GFI_22_28d', 'GBW_28', 'GBW_21', 'Mort_Wt_D28', 'Cull_Wt_D28', 'Fnl_Birds_Day28'),
                    ('cFCR29_35', 'GFI_29_35d', 'GBW_35', 'GBW_28', 'Mort_Wt_D35', 'Cull_Wt_D35', 'Fnl_Birds_Day35'),
                    ('cFCR36_42', 'GFI_36_42d', 'GBW_42', 'GBW_35', 'Mort_Wt_D42', 'Cull_Wt_D42', 'Fnl_Birds_Day42')]
    
    mort_columns = [('Mort7', 'Init_Birds_Day1', 'Fnl_Birds_Day7'),
                    ('Mort14', 'Init_Birds_Day1', 'Fnl_Birds_Day14'),
                    ('Mort21', 'Init_Birds_Day1', 'Fnl_Birds_Day21'),
                    ('Mort28', 'Init_Birds_Day1', 'Fnl_Birds_Day28'),
                    ('Mort35', 'Init_Birds_Day1', 'Fnl_Birds_Day35'),
                    ('Mort42', 'Init_Birds_Day1', 'Fnl_Birds_Day42')]

    # Función auxiliar para verificar columnas y calcular
    def verify_and_calculate(df, column_set, formula):
        for col_set in column_set:
            new_col = col_set[0]
            required_cols = col_set[1:]
            # Verificar si todas las columnas requeridas están en el DataFrame
            if all(col in df.columns for col in required_cols):
                df[new_col] = formula(df, *required_cols)
            else:
                missing_cols = [col for col in required_cols if col not in df.columns]
                print(f"Cannot calculate '{new_col}': missing columns {missing_cols}")

    # Definir fórmulas para cada cálculo
    verify_and_calculate(df, bw_columns, lambda df, gbw, birds: df[gbw] / df[birds])
    verify_and_calculate(df, fcr_columns, lambda df, fi, gbw, gbw_init, birds: (df[fi] / (df[gbw] - df[gbw_init])))
    verify_and_calculate(df, cfcr_columns, lambda df, fi, gbw, gbw_init, mort_wt, cull_wt, birds: (df[fi] / (df[gbw] - df[gbw_init] + df[mort_wt] + df[cull_wt])))
    verify_and_calculate(df, mort_columns, lambda df, ini_birds, fnl_birds: ((df[ini_birds] - df[fnl_birds]) / df[ini_birds]) * 100)
    return df
```
